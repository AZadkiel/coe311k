```python
import numpy as np
import matplotlib.pyplot as plt

#loop(range) your step size to go from .01 to 1, steps of .05
#Plot each step-size run, keep an eye on how the error is related to the step-size, print the error(current run vs first run)
#for each point, and the total accumulated error for each run.

# Define parameters

# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y

realY = lambda x : np.exp(-2 * x)

for h in np.arange(0.01, 1.01, 0.05): #step size
    x_start, x_end = 0, 2
    num_steps = int((x_end - x_start) / h)
    
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    
    # Initial condition
    y_values[0] = 1  # y(0) = 1

    #Implement Euler's method
    for i in range(num_steps):

        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    trueY = realY(x_values)
    currentError = np.abs(y_values - trueY)
    cumulativeError = np.sum(currentError)
    
    print(f" Current Step: {h:<0.4f} | Cumulative Error: {cumulativeError:<0.4f}")

    
xEuler = np.linspace(x_start, x_end, 100)
plt.plot(xEuler, realY(xEuler), 'k--', label="Exact Solution (e^-2x)")

plt.plot(x_values, y_values, label="Euler's Method Approximation")
plt.xlabel('x')
plt.ylabel('y')
plt.title("Approximation of dy/dx = -2y using Euler's Method")
plt.legend()
plt.show()
```

The larger the step size, the larger the cumulative error
