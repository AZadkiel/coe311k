```python
import numpy as np
from timeit import default_timer as timer
import matplotlib.pyplot as plt

def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C

def matrixMaker(nRows, nCols):
    M = np.random.randint(10, size = (nRows, nCols))
    return M

def matrixDoubleMultTime(initialSize, pairs):
    naiveTime = []
    numpyTime = []
    for i in range (pairs):
        M = np.random.randint(10, size = ((initialSize*(2**i)), (initialSize*(2**i))))
        Z = np.random.randint(10, size = ((initialSize*(2**i)), (initialSize*(2**i))))
        
        start = timer()
        naive_matrix_multiplication(M, Z)
        end = timer()
        naiveTime.append(end - start)
                              
        start = timer()
        np.matmul(M, Z)
        end = timer()
        numpyTime.append(end - start)
                          
    return naiveTime, numpyTime

naiveResults, numpyResults = matrixDoubleMultTime(3, 8)
print(naiveResults)
print(numpyResults)
matrixNumber = list(range(1, 9))

plt.plot(matrixNumber, naiveResults, label='Naive', marker='o', color='red',)
plt.plot(matrixNumber, numpyResults, label='NumPy', marker='s', color='blue')
plt.show()
```

Based on the graph above, Numpy is much more optimized compared to the Naive multiplication
