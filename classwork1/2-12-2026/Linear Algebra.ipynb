```python
import numpy as np

v1 = np.array([1, 2, 3]) #A 3-dimensional vector; and don't forget paranthesis
v2 = np.array([4, 5, 6])

print(v1)
print(v2)
```

```python
v_add = v1 + v2
v_sub = v1 - v2
print("v1 + v2:", v_add)
print("v1 - v2:", v_sub)
```

```python
scalar = 2
v_scaled = scalar*v1

print(v_scaled)
```

```python
dot_product = np.dot(v1, v2)

print(dot_product)
```

```python
magnitude_v1 = np.linalg.norm(v1)

print(magnitude_v1)
```

```python
# Creating a 2x3 matrix
A = np.array([[1, 2, 3], [4, 5, 6]]) # don't forget double bracket
print("Matrix A:\n", A)

# Creating a 3x3 matrix
B = np.array([[7, 8, 9], [10, 11, 12], [13, 14, 15]])
print("Matrix B:\n", B)
```

```python
C = np.array([[1, 2, 3], [4, 5, 6]])
matrix_add = A + C
matrix_sub = A - C

print("A + C =\n", matrix_add)
print("A - C =\n", matrix_sub)
```

```python
matrix_mult = np.dot(A, B)
print("A * B:/n", matrix_mult)
```

```python
transpose_A = A.T
print("Transpose of A:\n", transpose_A)
```

```python
A = np.array([[2, 3], [4, -1]])
b = np.array ([5, 3])
print(A)
print(b)
x = np.linalg.solve(A, b)
print(x)
```

```python
A = np.array([[2, 1],[1, 2]])
e_value, e_vectors = np.linalg.eig(A)

print("eigenvalue: \n:", e_value)
print("eigenvectors:\n", e_vectors)
```

```python

A = np.array([[4, 7], [2, 6]])
A_inv = np.linalg.inv(A)

print(A)
print("Inverse of A:\n", A_inv)
```

```python
identity = np.dot(A, A_inv)
print(identity)
```

```python
# Define a matrix
A = np.array([[1, 2], [3, 4], [5, 6]])
# Perform SVD
U, sigma, Vt = np.linalg.svd(A)
print("U matrix:\n", U)
print("Sigma values:", sigma)
print("V^T matrix:\n", Vt)
```
